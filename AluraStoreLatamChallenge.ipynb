#Extra – Análisis geográfico
## Análisis geográfico

En esta etapa opcional se exploraron las **coordenadas geográficas (latitud y longitud)** de las transacciones realizadas en cada tienda.  
El objetivo fue visualizar la ubicación de las compras y detectar posibles patrones regionales que influyan en las ventas o en la satisfacción de los clientes.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(tienda['lon'], tienda['lat'], label='Tienda 1', alpha=0.5)
plt.scatter(tienda2['lon'], tienda2['lat'], label='Tienda 2', alpha=0.5)
plt.scatter(tienda3['lon'], tienda3['lat'], label='Tienda 3', alpha=0.5)
plt.scatter(tienda4['lon'], tienda4['lat'], label='Tienda 4', alpha=0.5)

plt.title('Distribución geográfica de ventas por tienda')
plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.legend()
plt.grid(True)
plt.show()
```

### Interpretación del análisis geográfico

El gráfico de dispersión muestra la distribución geográfica de las ventas registradas por las cuatro tiendas.  
Las coordenadas de latitud y longitud evidencian que las transacciones se concentran principalmente en las principales regiones urbanas del país, como Bogotá, Medellín y Cali, con una distribución muy similar entre las tiendas.  

No se observan diferencias significativas en la ubicación de los clientes que expliquen los contrastes en ventas o calificaciones.  
Por lo tanto, se concluye que la variable geográfica no influye directamente en el desempeño: los resultados dependen de factores internos, como la gestión comercial, las categorías de productos y la experiencia de compra.

##Índice de desempeño y recomendación final


```python
#Metricas por tienda
import pandas as pd

resumen = pd.DataFrame({
    'Tienda': ['Tienda 1','Tienda 2','Tienda 3','Tienda 4'],
    'Ingreso': [
        tienda['Precio'].sum(),
        tienda2['Precio'].sum(),
        tienda3['Precio'].sum(),
        tienda4['Precio'].sum()
    ],
    'Calificación': [
        tienda['Calificación'].mean(),
        tienda2['Calificación'].mean(),
        tienda3['Calificación'].mean(),
        tienda4['Calificación'].mean()
    ],
    'EnvioProm': [
        tienda['Costo de envío'].mean(),
        tienda2['Costo de envío'].mean(),
        tienda3['Costo de envío'].mean(),
        tienda4['Costo de envío'].mean()
    ],
    'Ventas': [
        tienda['Producto'].count(),
        tienda2['Producto'].count(),
        tienda3['Producto'].count(),
        tienda4['Producto'].count()
    ]
})

resumen
```

```python
#Normalización + Índice de desempeño
df = resumen.copy()

for col in ['Ingreso', 'Calificación', 'Ventas']:
    df[col + '_Norm'] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())

df['EnvioProm_Norm'] = (df['EnvioProm'].max() - df['EnvioProm']) / (df['EnvioProm'].max() - df['EnvioProm'].min())
df['IndiceDesempeño'] = df[['Ingreso_Norm','Calificación_Norm','Ventas_Norm','EnvioProm_Norm']].mean(axis=1)
df_ordenado = df.sort_values('IndiceDesempeño', ascending=False).reset_index(drop=True)
df_ordenado[['Tienda','Ingreso','Calificación','EnvioProm','Ventas','IndiceDesempeño']]
```

```python
# Peor desempeño
df.loc[df['IndiceDesempeño'].idxmin(), ['Tienda','IndiceDesempeño']]
```

```python
tienda_cerrar = df.loc[df['IndiceDesempeño'].idxmin(), 'Tienda']
pd.DataFrame({'Recomendación': [ 'Vender ' + tienda_cerrar + ' (menor índice de desempeño).' ]})
```

## Recomendación para el Sr. Juan

### Introducción
El presente informe resume el análisis de desempeño de las cuatro tiendas de la cadena Alura Store, con el objetivo de determinar cuál de ellas debería ser vendida para que el Sr. Juan pueda invertir en un nuevo proyecto.

El estudio se basó en cinco indicadores clave:
- Facturación total  
- Categorías y productos más vendidos  
- Calificación promedio de clientes  
- Costo de envío promedio  
- Índice de desempeño general (combinando métricas normalizadas)

---

### Análisis de resultados

#### Facturación total
La **Tienda 1** registró el mayor nivel de ingresos con aproximadamente `$1.150.880.400`,  
mientras que la **Tienda 4** fue la de menor facturación con `$1.038.375.700`.  
Esto demuestra una diferencia importante en el rendimiento económico entre ambas.

#### Calificación promedio
Las calificaciones oscilaron entre `3.97` y `4.05`.  
Las **Tiendas 2 y 3** obtuvieron las mejores valoraciones por parte de los clientes,  
mientras que la **Tienda 1** presentó la calificación más baja.

#### Costo de envío promedio
El costo de envío fue un factor relevante:  
la **Tienda 4** presentó el valor promedio más bajo (`$23.459,46`),  
mientras que la **Tienda 1** registró el más alto (`$26.018,61`).  
Aunque la Tienda 4 posee costos de envío competitivos,  
su bajo volumen de ingresos y ventas afecta su rendimiento general.

#### Índice de desempeño general
Al combinar los indicadores de ingresos, calificación, ventas y costos de envío,  
se calculó un **Índice de Desempeño Normalizado**.  
Este índice permite comparar objetivamente la eficiencia global de cada tienda:

| Tienda | Índice de Desempeño |
|:-------|--------------------:|
| Tienda 3 | 0.75 |
| Tienda 2 | 0.71 |
| Tienda 1 | 0.50 |
| Tienda 4 | 0.32 |

---

### Conclusión
De acuerdo con los resultados obtenidos, la **Tienda 4** presenta el menor índice de desempeño (`0.32`),  
lo que refleja un rendimiento inferior en comparación con las demás tiendas.  

Si bien cuenta con costos de envío bajos, sus ingresos y ventas totales son los más reducidos,  
lo que disminuye su rentabilidad global.  

Por lo tanto, se recomienda **vender la Tienda 4**,  
ya que es la menos rentable del grupo y su desempeño general la posiciona como la mejor opción para desprenderse de ella.
